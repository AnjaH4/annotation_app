```python
import os
import numpy as np
from glob import glob
import json
from pathlib import Path

```

```python
fake_path = 'surveyapp/static/img/SBIs/sbi_fake/frames'
real_path = 'surveyapp/static/img/SBIs/sbi_real/frames'

n_frames = 4

image_list = []
label_list = []
fake_list = sorted(glob(fake_path+'/**'))
real_list = sorted(glob(real_path+'/**'))
filelist = []
#list_dict = json.load(open(f'/d/hpc/projects/FRI/borutb/DataBase/DeepFake/FaceForensics++/original_videos/{phase}.json','r'))
#list_dict = json.load(open(f'/home/anjah/Documents/mag/DeepFakes/SelfBlendedImages/data/ff++/original_videos/train.json', 'r'))
#list_dict = json.load(open(f'/shared/workspace/lrv/datasets/train.json', 'r'))

image_dir = Path('surveyapp/static/img/SBIs/')
all_images = list(image_dir.rglob('*.png'))


#print(fake_list)

# Optional: convert to string paths
all_images = [str(p) for p in all_images]
all_images = [i.split('/')[-2] for i in all_images]
filelist = []
print(len(all_images))

filelist = all_images
print(filelist[:5])

fake_list = [i for i in fake_list if os.path.basename(i)[:3] in filelist]
real_list = [i for i in real_list if os.path.basename(i)[:3] in filelist]

print(len(fake_list))
print(len(real_list))
for i in range(len(real_list)):
	images_temp=sorted(glob(real_list[i]+'/*.png'))
	if n_frames<len(images_temp):
		images_temp=[images_temp[round(i)] for i in np.linspace(0,len(images_temp)-1,n_frames)]
	image_list+=images_temp
	label_list+=[0]*len(images_temp)

for i in range(len(fake_list)):
	images_temp=sorted(glob(fake_list[i]+'/*.png'))
	if n_frames<len(images_temp):
		images_temp=[images_temp[round(i)] for i in np.linspace(0,len(images_temp)-1,n_frames)]
	image_list+=images_temp
	label_list+=[1]*len(images_temp)

print(f'Number of images: {len(image_list)}')
print(f'Number of labels: {len(label_list)}')

# image_list=image_list[:10]
# label_list=label_list[:10]
#21
#30,
```
